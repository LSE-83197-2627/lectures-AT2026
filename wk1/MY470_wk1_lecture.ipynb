## Computational Thinking

All knowledge can be thought as either declarative or imperative:

* **Daclarative knowledge** = statements of facts
    * "The squareroot of $x$ is a number $y$ such that $y \times y = x$
* **Imperative knowledge** = "how to" knowledge, recipes for deducing information
    * How can we compute the square root of a number?

> Computational Thinking is the thought processes involved in formulating a problem and expressing its solution in a way that a computer — human or machine — can effectively carry out.


Wing, Jeannette M. (2006). [Computational thinking](https://dl.acm.org/doi/pdf/10.1145/1118178.1118215). *Communications of the ACM*, 49(3), 33-35.

## Algorithms

An algorithm is a well-defined computational procedure that takes value(s) as input and produces value(s) as output.

  * "Recipe" or "instructions" for solving a well-defined computational problem
  * Consists of a sequence of simple steps, control flow, and a stopping rule
  * Can be specified in human language or programming language (or even as hardware design)
  

For example, **a guess-and-check algorithm to find the square root $y$ of a number $x$**, such that y \times y = x$

1. Start with a guess $g$
2. If $g \times g$ is close enough to $x$, stop and say that the answer is $g$
3. Otherwise create a new guess by averaging $g$ and $\frac{x}{g}$, i.e. $\frac{1}{2} \frac{g+x}{g}$
4. Using this new guess, which we again call $g$, repeat the process until $g \times g$ is close enough to $x$

## What Makes a Programming Languages?

If we want a computer to execute an algorithm, we need a way to instruct the computer to execute all involved steps in the right order.

A programming language is a formal language used to specify a set of instructions for a computer to execute.

Each programming language has

* Set of primitive constructs
* Syntax
* Static semantics
* Semantics

## What Makes a Programming Languages?

Analogy to natural languages:

| Programming language | Natural-language analogy |
|:----:|:------|
| Primitive Constructs | Words |
| Syntax | describes which strings of words constitute well-formed sentences |
| Static semantics | describes which sentences are meaningful |
| Semantics | defines the meaning of those sentences |

### Primitive Constructs in Programming Languages

Atomic building blocks of a language. In Python, primitives include:

* Literals

```python
470
```

```python
'MY'
```

* Infix operators

```python
470/3
```

* (→ as opposed to prefix and postfix operators)

| Prefix | Infix | Postfix
|:----:|:----:|:------|
| `+ 1 2` | `1 + 2` | `1 2 +` |

### Syntax in Programming Languages

* Rules for forming strings of characters and symbols
* Programming languages have strict syntax

* Syntactically valid sequence of primitives, e.g. `<literal>` `<infix operator>` `<literal>`:

```python
470 + 0.5
```

* Syntacticaly invalid sequence of primitives, e.g. `<literal>` `<literal>`, or `<literal>` `<infix operator>`:

```python
470 0.5
```

```python
470 +
```

### Static Semantics in Programming Languages

* Defines which syntactically valid strings have a meaning

```python
'MY'/470
```

→ syntactically well formed (`<literal>` `<infix operator>` `<literal>`), but not meaningful

### Semantics in Programming Languages

* The meaning associated with a syntactically correct string that has no static semantic errors
* Programming languages have clear semantics 
    * → if your code runs, it has **one and only one meaning**
    * → other than natural languages, programming languages allow **no room for ambiguity**

* **But this may not be the meaning the programmer had in mind!**

## The `print` Function in Python

```python
print('The')
print('The', 'winning', 'number', 'is', 7, '.')
print('The winning number is '+ str(7) + '.')
```

## Debugging Systematically

1. Compare input in successful and failing runs
2. Formulate a hypothesis
3. Design an experiment to test the hypothesis; use `print()`
4. Keep record of your experiment
5. Repeat
