## Basics of Coding in Jupyter Notebooks

<div style="font-size: 0.9em;">

* Code cells are run by pressing `CTRL+ENTER` or by clicking the `Run Cell` button
* The output of the last line of code in a cell is displayed below the cell
* Any _variables_ you create are shared between cells in the same notebook, and can be explored by clicking the `Variables` button in the toolbar
* If code is taking too long to run, you can stop it by pressing the interrupt button
* If you want to restart the kernel (i.e. clear all variables from memory), you can do so by pressing the `Restart` button
  * This can help if you are encountering bugs that you can't figure out
* `Run All` runs all cells in the notebook from top to bottom. Variables from earlier cells can then be used in later cells. Your code should run from top to bottom without any errors.

</div>
<div style="box-sizing: border-box; margin-top: 18px; padding: 12px 18px; border-radius: 8px; background-color: #f7efe6; font-family: system-ui; line-height: 1.45;"><b>Before submitting assignments, always <code>Restart Kernel</code> and <code>Run All</code> to make sure your code runs fully from a clean slate.</b></div>

## Useful Bash Commands

| Command | What it does |
|:--|:--|
| `pwd` | Print current working directory |
| `cd Path/to/directory` | Change current working directory |
| `cd ..` | Go back to the parent directory of the current one |
| `cd ~` | Go back to your home directory |
| `mkdir dirname` | Create a new directory |
| `ls` | Print a list of files and subdirectories |
| `python` | Launch a Python interpreter (type `exit()` to stop and go back to bash) |

## Change Your Default Text Editor for Git

You can use your favorite editor by customizing the Git default editor.

For example, you can use [**Nano**](https://www.nano-editor.org/). It is much easier to use than Vim: `Ctrl+o` to save and `Ctrl+x` to close.

To set Nano as the default editor for your commit messages, run the following:

```
git config --global core.editor "nano"
```

Nano comes pre-installed with Linux and OS. For Windows, download and install [**Nano-win**](https://github.com/mcandre/nano-win).
