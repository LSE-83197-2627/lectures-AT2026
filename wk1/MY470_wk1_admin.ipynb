# Why Do Social Scientists Need Computer Programming?

## Why Do Social Scientists Need Computer Programming?

* **Collect existing data**
  * Web scraping and APIs
  * Digital trace data (online platforms, phones, sensors)


* **Generate new data**
  * Online surveys and survey experiments
  * Online platform experiments
  * Computational models and simulations

* **Turn unstructured data into measures**
  * Text as data (NLP, word embeddings)
  * Audio, video, and images as data
  * Satellite and remote sensing data
  * Machine learning classification

* **Manage and analyze complex data**
  * Large data
  * Linking multiple data sources
  * Network analysis

## Why Do Social Scientists Need Computer Programming?

Proficiency in programming will enable you to

* Be autonomous and work independently
* Learn from and collaborate with engineers and scientists
* Generate and share reproducible workflows

## Coding vs. Programming

Some of you may have experience in **coding**:

* Scripts that call existing functions, e.g. `lm()` or `ggplot()` in R, or `regress` in Stata
* Someone else designed the tool, and you apply it

**Programming** goes beyond coding. It means designing your own programs:

* Break a research problem into smaller tasks
* Work out the step-by-step logic to solve each one
* Turn that logic into your own functions and tools
* Test, debug, and adapt your code

_Programming = "the designing, writing, modifying, and testing of programs"_ ([ANSDIT](https://www.incits.org/html/ext/ANSDIT/p3.htm))

## Assessment Criteria (Problem Sets)

  * The code runs and does what it is expected to
  * The code is written using the concepts, paradigms, and best practices covered in the course
    1. 📖 **Legibility** (Formatting of the code, comments, naming conventions)
    2. ⚙️ **Modularity** (Breaking the problem into functions, avoiding repetition, clear inputs and outputs)
    3. ⚡️ **Optimization** (Efficient algorithms, avoiding unnecessary computation - sometimes in tension with legibility)

## Collaboration, Plagiarism, and AI Policy

* Problem sets and the quiz are individual unless we instruct you otherwise
* For individual problem sets:
  * You are not allowed to discuss solutions with peers, friends, family, roommates, etc.
  * You are not allowed to show your code or view others' code (including solutions shared by previous students)
  * You are not allowed to ask questions about the problem sets on Q&A sites such as Stack Overflow
  * **You are not allowed to use AI tools** such as ChatGPT and Copilot unless we instruct you otherwise

* You can search for general advice online (e.g. on Stack Overflow) but always give credit in comments if you borrow large chunks of code or non-standard solutions  
* You can use the forum "Clarifying Questions about Assignments" on **Moodle** to ask questions about the problem set instructions (but not about the solutions!)

## Why are AI tools prohibited?

Without a doubt, when you are an employee (industry or academia) in the future, you will use GenAI tools to help you programming. **But:**

* You are not here to get a job done, but to develope skills
* This is a foundational class and you are learning foundational skills
* Computational thinking and computational problem solving is more than writing code, but coding helps learn computational thinking and problemsolving
* GenAI won't help you in learning those foundational skills
